```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

Choose a stock:

```python
ticker_symbol = "AAPL"  #example, AAPL
start_date = "2020-01-01"     #format: "2020-01-01"
end_date = "2024-01-01"
```

```python
ticker = yf.Ticker(ticker_symbol)
historical_data = ticker.history(start=start_date, end=end_date)
df = pd.DataFrame(historical_data)
```

The dataframe called df contains a table of data related to the stock

```python
df    #print(df)
```

Run this next box to graph the price data from above

```python
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['Close'], label='Close Price', color='blue')
plt.title('Historical Stock Price Data for ' + ticker_symbol)
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.show()
```
